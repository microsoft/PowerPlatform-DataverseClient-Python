## Async-only SDK in Jupyter — what sync users experience

Jupyter runs all cells inside a **persistent event loop**. This notebook shows what happens when a sync user tries to use an async-only SDK.

```python
import asyncio

# Simulated async-only SDK method (no sync client exists)
async def sdk_create_record(table: str, data: dict) -> str:
    await asyncio.sleep(0)  # yields control, like a real network call
    return f"fake-guid-for-{data.get('name', 'unknown')}"
```

### Attempt 1 — call without `await` (forgetting to await)

A user unfamiliar with async might call the function directly.

```python
result = sdk_create_record("account", {"name": "Contoso"})
print(f"Return value: {result}")
print(f"Type:         {type(result)}")
print()
print("The coroutine was never executed — no network call happened.")
```

### Attempt 2 — use `asyncio.run()` (the "sync" workaround)

A sync user who learns about `asyncio.run()` tries to use it to avoid writing `async def`.

```python
# This is what a sync user would try after reading the docs
record_id = asyncio.run(sdk_create_record("account", {"name": "Contoso"}))
print(f"Created: {record_id}")
```

### Attempt 3 — use `await` (correct, but requires `async def`)

The only way to call the SDK in Jupyter is with `await` — which works because Jupyter supports top-level `await`.

```python
# This works — Jupyter supports top-level await
record_id = await sdk_create_record("account", {"name": "Contoso"})
print(f"Created: {record_id}")
print()
print("Works — but the user must learn async/await just to use the SDK.")
```
